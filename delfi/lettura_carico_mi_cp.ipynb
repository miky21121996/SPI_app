# Loads in Milan 

```python
import numpy as np
import pandas as pd
pd.options.plotting.backend = "plotly"
```

## Reading separate years files and combining them together

### File parsing

```python
loads_2021 = pd.read_csv("Consumi_CPMI_2021.csv", sep=";", parse_dates=["DATA_CORRETTA"], dayfirst=False)
loads_2022 = pd.read_csv("Consumi_CPMI_2022.csv", sep=";", parse_dates=["DATA_CORRETTA"], dayfirst=False)
loads_2023 = pd.read_csv("Consumi_CPMI_2023.csv", sep=";", parse_dates=["DATA_CORRETTA"], dayfirst=False)
```

```python
loads_2021
```

```python
loads_2022
```

```python
loads_2023
```

### Timestamp unification

```python
loads_2023['DATA_CORRETTA'] = pd.to_datetime(loads_2023['DATA_CORRETTA'], utc=True).dt.tz_convert('Europe/Rome').dt.tz_localize(None)
```

### Concat in single dataframe

```python
loads = pd.concat([loads_2021, loads_2022, loads_2023], ignore_index=True)
loads = loads.sort_values(by=["DATA_CORRETTA"], ascending=True)
loads = loads.reset_index(drop=True)
loads
```

```python
loads.columns
```

### Data conversion and fill of missing

```python
# List of columns to exclude from conversion
exclude_columns = ["DATA_CORRETTA", "OraLegale", "fascia"]

# Convert all other columns to float
loads[loads.columns.difference(exclude_columns)] = loads[loads.columns.difference(exclude_columns)].apply(pd.to_numeric, errors='coerce')

loads
```

```python
# percentage of missing
loads.isna().sum() / len(loads) * 100
```

```python
# remove ora legale and fascia
loads = loads.drop(columns=["OraLegale", "fascia"], axis=1)
```

```python
# interpolate nan values
loads = loads.interpolate().ffill().bfill()

loads
```

```python
loads.set_index("DATA_CORRETTA")["VIT2R"].plot()
```

```python
loads.set_index("DATA_CORRETTA")["LATR2V"].plot()
```

### Output not aggregated

```python
# Resample to hourly
loads_hourly = loads.set_index("DATA_CORRETTA").resample("h").sum()

# convert to MWh
loads_hourly = loads_hourly / 1000

loads_hourly
```

```python
loads_hourly.reset_index().rename(columns={"DATA_CORRETTA": "orario"}).to_csv("load_CP_MI_separate.csv", index=False)
```

### Output aggregated (sum of all primary substations)

```python
loads["TOTALE"] = loads[loads.columns.difference(exclude_columns)].sum(axis=1)
# where totale=0 impute linearly
loads["TOTALE"] = loads["TOTALE"].replace(0, np.nan).interpolate()
loads.set_index("DATA_CORRETTA")["TOTALE"].plot()
```

```python
# Resample to hourly
sum_of_loads = loads[["DATA_CORRETTA", "TOTALE"]].set_index("DATA_CORRETTA").resample("h").sum()

# convert to MWh
sum_of_loads = sum_of_loads / 1000

sum_of_loads
```

```python
sum_of_loads.plot()
```

```python
sum_of_loads.reset_index().rename(columns={"DATA_CORRETTA": "orario", "TOTALE": "load_mwh"}).to_csv("load_CP_MI_sum.csv", index=False)
```
